```python
import numpy as np

from sklearn.linear_model import LinearRegression


X = np.arange(1, 11).reshape(-1, 1)

Y = np.array([12, 15, 17, 20, 23, 25, 28, 30, 33, 35])


model = LinearRegression()

model.fit(X, Y)


Y_pred = model.predict(X)

residuals = Y - Y_pred


print("Slope:", round(model.coef_[0], 4))

print("Intercept:", round(model.intercept_, 4))

print("Predicted values:", np.round(Y_pred, 4))

print("Residuals:", np.round(residuals, 4))
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(7, 4))

plt.scatter(Y_pred, residuals)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Fitted Values")

plt.ylabel("Residuals")

plt.title("Residual Plot")

plt.show()
```

```python
import scipy.stats as stats

import matplotlib.pyplot as plt


plt.figure(figsize=(7, 4))

stats.probplot(residuals, dist="norm", plot=plt)

plt.title("Normal Probability Plot of Residuals")

plt.show()
```

```python
X = np.arange(1, 11).reshape(-1, 1)

Y = np.array([10, 13, 15, 18, 20, 23, 25, 28, 30, 33])


model = LinearRegression()

model.fit(X, Y)


print("Regression coefficient:", round(model.coef_[0], 4))

print("Intercept:", round(model.intercept_, 4))

print("Empirical model:")

print("Y =", round(model.coef_[0], 4), "* X +", round(model.intercept_, 4))


print("Predicted values:")

print(np.round(model.predict(X), 4))
```
